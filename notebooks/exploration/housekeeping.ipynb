```python
import numpy as np
import matplotlib.pyplot as plt

def compute_azimuth_detailed(pos, vel, target_unit, debug=False):
    """
    Compute azimuth with detailed debugging information
    """
    # Satellite coordinate system
    suz = -pos / np.linalg.norm(pos)
    
    # Satellite x-axis
    vel_proj = np.dot(vel, suz) * suz
    sux_unnorm = vel - vel_proj
    sux_norm = np.linalg.norm(sux_unnorm)
    
    if debug:
        print(f"  suz (nadir): {suz}")
        print(f"  vel: {vel}")
        print(f"  vel_proj: {vel_proj}")
        print(f"  sux_unnorm: {sux_unnorm}")
        print(f"  sux_norm: {sux_norm}")
    
    if sux_norm < 1e-10:
        if debug:
            print("  WARNING: sux_norm is very small - velocity parallel to nadir!")
        return np.nan, None, None, None
    
    sux = sux_unnorm / sux_norm
    
    # Satellite y-axis
    suy = np.cross(suz, sux)
    
    # North direction in satellite frame
    uz = np.array([0, 0, 1])
    uz_proj = np.dot(uz, suz) * suz
    north_unnorm = uz - uz_proj
    north_norm = np.linalg.norm(north_unnorm)
    
    if debug:
        print(f"  uz: {uz}")
        print(f"  uz_proj: {uz_proj}")
        print(f"  north_unnorm: {north_unnorm}")
        print(f"  north_norm: {north_norm}")
    
    if north_norm < 1e-10:
        if debug:
            print("  WARNING: north_norm is very small - satellite at pole!")
        return np.nan, sux, suy, suz
    
    north = north_unnorm / north_norm
    
    # Delta phi
    north_dot_suy = np.dot(north, suy)
    north_dot_sux = np.dot(north, sux)
    delta_phi = np.arctan2(north_dot_suy, north_dot_sux)
    
    # Target azimuth
    t_dot_sux = np.dot(target_unit, sux)
    t_dot_suy = np.dot(target_unit, suy)
    particle_phi = np.arctan2(t_dot_suy, t_dot_sux)
    
    if debug:
        print(f"  north: {north}")
        print(f"  north_dot_sux: {north_dot_sux}, north_dot_suy: {north_dot_suy}")
        print(f"  delta_phi: {np.degrees(delta_phi):.1f}°")
        print(f"  t_dot_sux: {t_dot_sux}, t_dot_suy: {t_dot_suy}")
        print(f"  particle_phi: {np.degrees(particle_phi):.1f}°")
    
    # Apply correction
    azimuth = -(particle_phi - delta_phi)
    azimuth = np.mod(azimuth, 2 * np.pi)
    
    if debug:
        print(f"  raw azimuth: {np.degrees(-(particle_phi - delta_phi)):.1f}°")
        print(f"  final azimuth: {np.degrees(azimuth):.1f}°")
    
    return np.degrees(azimuth), sux, suy, suz

def test_satellite_azimuth_detailed():
    """
    Enhanced test with better test cases and debugging
    """
    
    print("Enhanced Satellite Azimuth Test")
    print("="*50)
    
    # Better test cases
    test_cases = [
        {
            'name': 'Equator Satellite (0°N, 0°E)',
            'pos': np.array([6371e3, 0, 0]),  # On equator
            'vel': np.array([0, 7500, 0])     # Moving east
        },
        {
            'name': 'Equator Satellite (0°N, 90°E)', 
            'pos': np.array([0, 6371e3, 0]),  # On equator, 90° east
            'vel': np.array([-7500, 0, 0])    # Moving west (opposite to pos)
        },
        {
            'name': 'Mid-latitude Satellite (45°N)',
            'pos': np.array([4502e3, 0, 4502e3]),  # 45° north
            'vel': np.array([0, 7500, 0])          # Moving east
        },
        {
            'name': 'Near-polar Satellite (80°N)',
            'pos': np.array([1108e3, 0, 6278e3]),  # 80° north
            'vel': np.array([0, 7500, 0])          # Moving east
        }
    ]
    
    # Test targets with clear expected results
    targets = {
        'North Pole': np.array([0, 0, 1]),
        'South Pole': np.array([0, 0, -1]),
        'Equator at 0°': np.array([1, 0, 0]),
        'Equator at 90°E': np.array([0, 1, 0]),
        'Equator at 180°': np.array([-1, 0, 0]),
        'Equator at 270°E': np.array([0, -1, 0])
    }
    
    for test_case in test_cases:
        print(f"\n{test_case['name']}:")
        print(f"Position: {test_case['pos']/1e6} Mm")
        print(f"Velocity: {test_case['vel']/1e3} km/s")
        
        for target_name, target in targets.items():
            print(f"\n  Target: {target_name}")
            azimuth, sux, suy, suz = compute_azimuth_detailed(
                test_case['pos'], test_case['vel'], target, debug=False
            )
            
            if np.isnan(azimuth):
                print(f"    Azimuth: NaN (degenerate case)")
            else:
                print(f"    Azimuth: {azimuth:6.1f}°")
                
                # Also compute elevation angle for context
                if suz is not None:
                    cos_theta = np.dot(target, suz)
                    elevation = 90 - np.degrees(np.arccos(np.abs(cos_theta)))
                    print(f"    Elevation: {elevation:6.1f}°")

def test_coordinate_system():
    """
    Test the coordinate system construction itself
    """
    print("\n" + "="*50)
    print("Coordinate System Test")
    print("="*50)
    
    # Simple equatorial case
    pos = np.array([6371e3, 0, 0])  # On equator at 0° longitude
    vel = np.array([0, 7500, 0])    # Moving east
    
    print("Test case: Satellite at (6371 km, 0, 0), velocity (0, 7.5, 0) km/s")
    print("\nDetailed computation:")
    
    target = np.array([0, 0, 1])  # North pole
    azimuth, sux, suy, suz = compute_azimuth_detailed(pos, vel, target, debug=True)
    
    print(f"\nFinal coordinate system:")
    print(f"SUX (velocity direction): {sux}")
    print(f"SUY (cross product):      {suy}")
    print(f"SUZ (nadir):             {suz}")
    
    # Check orthogonality
    if sux is not None and suy is not None and suz is not None:
        print(f"\nOrthogonality check:")
        print(f"SUX · SUY = {np.dot(sux, suy):.6f} (should be ~0)")
        print(f"SUY · SUZ = {np.dot(suy, suz):.6f} (should be ~0)")
        print(f"SUZ · SUX = {np.dot(suz, sux):.6f} (should be ~0)")
        
        print(f"\nMagnitude check:")
        print(f"|SUX| = {np.linalg.norm(sux):.6f} (should be 1)")
        print(f"|SUY| = {np.linalg.norm(suy):.6f} (should be 1)")
        print(f"|SUZ| = {np.linalg.norm(suz):.6f} (should be 1)")

def analyze_original_algorithm():
    """
    Analyze potential issues in the original algorithm
    """
    print("\n" + "="*50)
    print("Algorithm Analysis")
    print("="*50)
    
    print("""
Potential Issues Identified:

1. POLAR SATELLITE PROBLEM:
   - When satellite is at/near north pole, the "north" vector becomes ill-defined
   - uz (north pole) and suz (nadir) are nearly parallel
   - north = uz - uz.dot(suz)*suz approaches zero vector
   - This causes division by zero in normalization

2. COORDINATE SYSTEM DEFINITION:
   - The algorithm assumes orbital motion (velocity perpendicular to nadir)
   - For polar orbits or when velocity has significant radial component,
     this may not hold perfectly

3. REFERENCE FRAME CONVENTION:
   - The deltaPhi correction aligns the satellite frame with geographic north
   - This may not be the intended reference for all applications

Suggestions:
   - Add checks for degenerate cases (satellite at poles)
   - Consider alternative reference directions for polar regions
   - Validate the coordinate system is right-handed and properly oriented
    """)

test_satellite_azimuth_detailed()
test_coordinate_system()
analyze_original_algorithm()
```
